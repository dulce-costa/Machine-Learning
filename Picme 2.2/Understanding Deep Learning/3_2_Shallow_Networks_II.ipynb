```python
# Imports math library
# PT: Importa a biblioteca de matemática
import numpy as np
# Imports plotting library
# PT: Importa a biblioteca de gráficos
import matplotlib.pyplot as plt
```

```python
# Code to draw 2D function -- read it so you know what is going on, but you don't have to change it
# PT: Código para desenhar uma função 2D -- leia para saber o que está acontecendo, mas não é necessário alterá-lo
def draw_2D_function(ax, x1_mesh, x2_mesh, y):
    pos = ax.contourf(x1_mesh, x2_mesh, y, levels=256 ,cmap = 'hot', vmin=-10,vmax=10.0)
    ax.set_xlabel('x1');ax.set_ylabel('x2')
    levels = np.arange(-10,10,1.0)
    ax.contour(x1_mesh, x2_mesh, y, levels, cmap='winter')

# Plot the shallow neural network.  We'll assume input in is range [0,10],[0,10] and output [-10,10]
# PT: Plota a rede neural rasa. Vamos assumir que a entrada está no intervalo [0,10],[0,10] e a saída em [-10,10]
import matplotlib.pyplot as plt


def plot_neural_2_inputs(
    x1,
    x2,
    y,
    pre_1,
    pre_2,
    pre_3,
    act_1,
    act_2,
    act_3,
    w_act_1,
    w_act_2,
    w_act_3,
):
    fig, ax = plt.subplots(3, 3)
    fig.set_size_inches(10, 8.5)

    # Mapeamento dos eixos, dados e títulos
    plots = [
        (ax[0, 0], pre_1, "Preactivation"),
        (ax[0, 1], pre_2, "Preactivation"),
        (ax[0, 2], pre_3, "Preactivation"),
        (ax[1, 0], act_1, "Activation"),
        (ax[1, 1], act_2, "Activation"),
        (ax[1, 2], act_3, "Activation"),
        (ax[2, 0], w_act_1, "Weighted Act"),
        (ax[2, 1], w_act_2, "Weighted Act"),
        (ax[2, 2], w_act_3, "Weighted Act"),
    ]

    for axis, data, title in plots:
        im = draw_2D_function(axis, x1, x2, data)
        axis.set_title(title)

        # Captura o objeto gráfico retornado ou obtém o objeto desenhado no eixo (collections/images)
        mappable = (
            im
            if im is not None
            else (axis.collections[0] if axis.collections else axis.images[0])
        )
        fig.colorbar(mappable, ax=axis)

    fig.tight_layout(pad=3.0)
    plt.show()

    # Figura da saída da rede
    fig, ax = plt.subplots()
    im = draw_2D_function(ax, x1, x2, y)
    ax.set_title("Network output, $y$")
    ax.set_aspect(1.0)

    mappable = (
        im
        if im is not None
        else (ax.collections[0] if ax.collections else ax.images[0])
    )
    fig.colorbar(mappable, ax=ax)

    plt.show()
```

```python
# Define the Rectified Linear Unit (ReLU) function
# PT: Define a função ReLU (Unidade Linear Retificada)
def ReLU(preactivation):
  activation = preactivation.clip(0.0)
  return activation
```

```python
# Define a shallow neural network with, two input, one output, and three hidden units
# PT: Define uma rede neural rasa com duas entradas, uma saída e três unidades ocultas
def shallow_2_1_3(x1,x2, activation_fn, phi_0,phi_1,phi_2,phi_3, theta_10, theta_11,\
                  theta_12, theta_20, theta_21, theta_22, theta_30, theta_31, theta_32):
  
  pre_1 = theta_10 + theta_11 * x1 + theta_12 * x2
  pre_2 = theta_20 + theta_21 * x1 + theta_22 * x2 
  pre_3 = theta_30 + theta_31 * x1 + theta_32 * x2

  # Pass these through the ReLU function to compute the activations as in
  # PT: Passa esses valores pela função ReLU para calcular as ativações como em
  # figure 3.8 d-f
  # PT: figura 3.8 d-f
  act_1 = activation_fn(pre_1)
  act_2 = activation_fn(pre_2)
  act_3 = activation_fn(pre_3)

  w_act_1 = phi_1 * act_1
  w_act_2 = phi_2 * act_2
  w_act_3 = phi_3 * act_3 

  y = phi_0 + w_act_1 + w_act_2 + w_act_3

  # Return everything we have calculated
  # PT: Retorna tudo o que foi calculado
  return y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3
```

```python
# Now lets define some parameters and run the neural network
# PT: Agora vamos definir alguns parâmetros e executar a rede neural
theta_10 =  -4.0 ;  theta_11 = 0.9; theta_12 = 0.0
theta_20 =  5.0  ; theta_21 = -0.9 ; theta_22 = -0.5
theta_30 =  -7  ; theta_31 = 0.5; theta_32 = 0.9
phi_0 = 0.0; phi_1 = -2.0; phi_2 = 2.0; phi_3 = 1.5

x1 = np.arange(0.0, 10.0, 0.1)
x2 = np.arange(0.0, 10.0, 0.1)
x1,x2 = np.meshgrid(x1,x2)  # https://www.geeksforgeeks.org/numpy-meshgrid-function/

# We run the neural network for each of these input values
# PT: Executamos a rede neural para cada um desses valores de entrada
y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3 = \
    shallow_2_1_3(x1,x2, ReLU, phi_0,phi_1,phi_2,phi_3, theta_10, theta_11, theta_12, theta_20, theta_21, theta_22, theta_30, theta_31, theta_32)
# And then plot it
# PT: E então plotamos o resultado
plot_neural_2_inputs(x1,x2, y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3)
```
